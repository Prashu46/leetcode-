```python
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        minprice=999
        max_profit=0
        for price in prices:
            minprice=min(minprice,price)
            max_profit=max(max_profit,price-minprice)

        return max_profit
```

```python
class Solution:
    def isIsomorphic(self, s: str, t: str) -> bool:
        hash1,hash2={},{}
        if len(s)!=len(t):
            return False
        for i in  range (len(s)):
            a1,b1=s[i],t[i]
            if ((a1 in hash1 and hash1[a1]!=b1) or (b1 in hash2 and hash2[b1]!=a1)):
                return False
            hash1[a1]=b1
            hash2[b1]=a1
        return True
```

```python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        intervals.sort(key=lambda i:i[0])
        res=[]
        n=len(intervals)
        start1=intervals[0][0]
        end1=intervals[0][1]
        for i in range (1,n):
            start2=intervals[i][0]
            end2=intervals[i][1]
            if end1>=start2:
                start1=start1
                end1=max(end1,end2)
                continue
            res.append([start1,end1])
            start1=start2
            end1=end2
        res.append([start1,end1])
        return res

```

```python
class Solution:
    def intervalIntersection(self, firstList: List[List[int]], secondList: List[List[int]]) -> List[List[int]]:
        res=[]
        i,j=0,0
        while(i<len(firstList)and j<len(secondList)):
            start1=firstList[i][0]
            end1=firstList[i][1]
            start2=secondList[j][0]
            end2=secondList[j][1]
            if start1<=start2:   #iga namge lisr1 [1,3] second [2,3] so sorted no issue [1,3] and [2,5] fail
                if end1>=start2:
                    s=max(start1,start2)
                    p=min(end1,end2)
                    res.append([s,p])
            else:
                if end2>=start1:   # so reverse the logic 
                    s=max(start1,start2)
                    p=min(end1,end2)
                    res.append([s,p])
            if end1<=end2: # we always check who leaves first  that is incremented 
                i+=1
            else:
                j+=1
        return res
                      
                 
```

```python
class Solution:
    def removeDuplicates(self, s: str) -> str:
        stack=[]
        for i in range(len(s)):
            if(len(stack)==0):
                stack.append(s[i])
            elif stack[-1]==s[i]:
                stack.pop()
            else:
                stack.append(s[i])
        return "".join(stack)
```

```python
class Solution:
    def search(self, nums, target):
        left, right = 0, len(nums) - 1

        while left <= right:
            mid = (left + right) // 2
            if nums[mid] == target:
                return True
            if nums[left] == nums[mid] == nums[right]:
                left += 1
                right -= 1
            elif nums[left] <= nums[mid]:
                if nums[left] <= target < nums[mid]:
                    right = mid - 1
                else:
                    left = mid + 1
            else:
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1
        return False
```

```python
class Solution:
    def nextGreaterElements(self, nums):
        stack = []
        stack.append(nums[-1])
        top = 0
        result = []
        result.append(-1)

        for i in range(len(nums)-2,-1,-1):
            
            while stack and stack[top] <= nums[i]:
                stack.pop()
                top -= 1

            if not stack:
                result.append(-1)
                stack.append(nums[i])
                top += 1
            else:
                result.append(stack[top])
                stack.append(nums[i])
                top += 1
                
        result = result[::-1]
        
        
        for i in range(1,len(result)):
            if result[i] == -1:
                for j in range(0,i):
                    if nums[j] > nums[i]:
                        result[i] = nums[j]
                        break
        return result




        
```

```python
class Solution:
    def firstUniqChar(self, s: str) -> int:
        hash = {}
        for char in s:
            hash[char] = hash.get(char, 0) + 1   
        for i, char in enumerate(s):
            if hash[char] == 1:
                return i
        return -1
```

```python
class Solution:
    def canConstruct(self, ransomNote: str, magazine: str) -> bool:
       ran,mag={},{}
       for i in (ransomNote):
           ran[i]=ran.get(i,0)+1
       for j in (magazine):
           mag[j]=mag.get(j,0)+1
       for char,count in ran.items():
           if mag.get(char,0)>count:
               return True      
       return False
```

```python
class Solution:
    def maxNumberOfBalloons(self, text: str) -> int:
        hash={}
        hash1={}
        res=999
        for i in text:
            hash[i]=hash.get(i,0)+1
        hash1["b"]=1
        hash1["a"]=1
        hash1["l"]=2
        hash1["o"]=2
        hash1["n"]=1
        for i,count in hash1.items():
            have=hash.get(i,0)
            res=min(res,have//count)
        return res
    
```

```python
class Solution:
    def longestPalindrome(self, s: str) -> int:
        hash1={}
        res=0
        odd=False
        for char in s:
            hash1[char]=hash1.get(char,0)+1
        for value in hash1.values():
            if value % 2==0:
                res+=value
            else:
                
                res+=value-1
                odd=True
        

        return res if odd==False else res+1
```

```python
class Solution:
    def fib(self, n):
        if (n==0):
            return 0
        elif(n==1):
            return 1
        else:
            return self.fib(n-1)+self.fib(n-2)
```

```python
class Solution:
    def peakIndexInMountainArray(self, arr: List[int]) -> int:
        l,r=0,len(arr)-1
        while(l<r):
            mid=(l+r)//2
            if arr[mid]>arr[mid+1]:
                r=mid
            else:
                l=mid+1
            return l
```

```python
class Solution:
    def pivotIndex(self, nums: List[int]) -> int:
        n=len(nums)
        prefix,suffix=(n) *[0],n* [0]
        
        for i in range(1,n):
            prefix[i]=prefix[i-1]+nums[i-1]
        for j in range (n-2,-1,-1):
            suffix[j]=suffix[j+1]+nums[j+1]
        for i in range(len(nums)):
          if(prefix[i]==suffix[i]):
           return i
```
